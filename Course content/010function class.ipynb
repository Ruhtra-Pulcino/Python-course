```python
numbers = (input("Type a sequence of numbers: "))

space_numbers = numbers.split()

last_list = []

for number in space_numbers:

    number = int(number)

    last_list.append(number)

def statistic(*args):

    media = sum(args) / len(args)
    bigger = max(args)
    smaller = min(args)

    return media, bigger, smaller

media, bigger, smaller = statistic(*last_list)

print(f"media: {media}")

print(f"bigger: {bigger}")

print(f"smaller: {smaller}")
```

```python
def statistic(*args):
    return sum(args) / len(args), max(args), min(args)

numbers = list(map(int, input("Type a sequence of numbers: ").split()))
 
media, bigger, smaller = statistic(*numbers)

print(f"media: {media}")

print(f"bigger: {bigger}")

print(f"smaller: {smaller}")
```
