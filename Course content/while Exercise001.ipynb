```python
"""

Exercise:

Create an algorithm that asks the user for a password and only exits the
While loop when the correct password is entered.

"""


password = "123456"

user = input("Type the password")

while user != password:

    print("Wrong password try again\n")

    user = input("Type the password")

print("Welcome to the system")
```

```python
import random

number = random.randint(1, 100)
attempts = 0

while True: 

    guess = int(input("Type your guess"))
    attempts += 1

    if guess < number :

        print("The secret number is bigger. ")

        print(f"Your attempt: {guess}\n")

    elif guess > number:

        print("The secret number is smaller. ")

        print(f"Your attempt: {guess}\n")

    else:

        print(f"Congrats you guessed the number: {number} in {attempts} attempts")

        break
```

```python
"""

Exercise:

Create an algorithm that reads *n* integers entered by the user,
and stops only when the user enters 0.

At the end, print the sum of all the entered numbers to the screen.

"""

sum_ = 0
right_number = 0
user = int(input("Type 0 to leave the system: "))

while user != right_number:

    sum_ += user

    user = int(input("Type 0 to leave the system: "))

print("You leave the system")
print(f"The sum of your attempts: {sum_}")
```

```python
"""

Exercise:

Create an algorithm that reads positive integers entered by the user
until the user enters a number less than 0. At the end, print the largest number entered.

"""

user_number = int(input("Type a number or a negative number to leave the system: "))

largest = -1

while user_number >= 0:

    user_number = int(input("Type a number or a negative number to leave the system: "))

    if user_number > largest:

        largest = user_number

print("You leave the system")
print(f"The largest number you typed: {largest}")
```

```python
"""
Exercise: ATM Simulation

Objective: Develop a Python program that simulates the
basic operations of an ATM. The user must be able to
check the balance, deposit money, withdraw money, and exit the program.

Requirements:

Check Balance:

- Upon selecting this option, the program must display the current account balance.

Deposit Money:

- The user must be able to enter an amount to deposit into the account.
- The amount must be positive.
- After a successful deposit, the account balance must be updated and a
confirmation message must be displayed.

Withdraw Money:

- The user must be able to enter an amount to withdraw from the account.
- The amount must be positive and must not exceed the current balance.
- After a successful withdrawal, the account balance must be
updated and a confirmation message must be displayed.

Exit:

- The user must be able to exit the program by choosing this option. 

Input Validation:

- The program must handle invalid inputs appropriately, displaying error messages when applicable. 

User Interface:

- The program must display a menu of options to the user and allow
the selection of actions to be performed. 
- The menu options must be presented within a loop, allowing
multiple operations until the user chooses to exit. 

ATM
1 - Check Balance
2 - Deposit Money
3 - Withdraw Money
4 - Exit
Choose an option (1-4):

Initial Balance:

- The account must start with an initial balance of R$ 1000.00.

Instructions:

- Start by initializing the balance and entering a loop that displays the options menu. 
- Implement each operation as described in the requirements. 
- Make sure to test all options and possible scenarios to ensure
the program works correctly.


Tip: You can use `if`, `elif`, and `else` statements along with a `while` loop
to handle user selections and keep the program running
until the exit option is chosen.

"""

balance = 1000

choose = int(input("Welcome to the ATM\n1 - Check balance\n2 - Deposit Money\n3 - Withdraw\n4 - Exit"))

while choose != 4:

    if choose == 1:

        print(f"Your balance: {balance}")

        choose = int(input("Welcome to the ATM\n1 - Check balance\n2 - Deposit Money\n3 - Withdraw\n4 - Exit"))

    elif choose == 2:

        deposit = int(input("How much do you want to deposit: "))

        balance += deposit

        print(f"Your balance: {balance}")

        choose = int(input("Welcome to the ATM\n1 - Check balance\n2 - Deposit Money\n3 - Withdraw\n4 - Exit"))

    elif choose == 3:

        withdraw = int(input("How much do you want to withdraw: "))

        balance -= withdraw

        print(f"Your balance: {balance}")

    elif choose == 4:

        print("You leave the ATM")

        break
```
