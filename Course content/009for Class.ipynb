```python
number = int(input("Type a number: "))

sum_ = 0

for i in range(1, number + 1):

    square = i**2
    sum_ += square

    print(f"Square of {i}: {square}")

print(f"\nThe sum of squares between 1 and {number}: {sum_}")
```

```python
height = 5

for i in range(height):

    space = height - i - 1
    text = 2 * i + 1
    print(" " * space + "*" * text)

height = 5

for i in range(height):
    space = i
    text = 2 * (height - i) - 1
    print(" " * space + "*" * text)
```
