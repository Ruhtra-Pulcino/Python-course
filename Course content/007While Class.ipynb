```python
number = 1

sum_ = 0

number2 = 10

print(f"Números pares entre 1 e {number2}: ")

while number <= number2:

    number += 1

    if number % 2 == 0:

        print(f"Number: {number}")

        sum_ += number

print(f"Sum of all the pair numbers between {number} and {number2}: {sum_}")
```
