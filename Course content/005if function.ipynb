```python
#if, elif, and else are Python keywords—not functions—that control which block of code runs based on conditions.

number1 = int(input("Digit a number: "))
number2 = int(input("Digit the second number: "))

if number1 > number2:

    print("The first number is bigger than the second")

elif number1 == number2:

    print("The numbers are equal")

else:

    print("The second number is bigger than the first")
```
