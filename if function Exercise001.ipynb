```python
import random

number = random.randint(1, 10)
print("Youre being challenge to guess a number between 1 and 10 good luck")
guess = int(input("Insert here your guess: "))

if number == guess:

    print(f"Congratulations the number was {number} and you got it right")

else:

    print(f"What a shame the number was {number} but you guessed {guess}")




```

```python
can_vote = 18

name = input("Hello dear voter, what is your name? ")
age = int(input(f"{name} type here your age, so that we can know if you are old enough to vote: "))

if age < can_vote:

    print("What a shame you can't vote yet")

elif age == can_vote or age > can_vote:

    print("Congrats you can vote!")
```

```python
name = input("Hello dear user whats your name: ")
note = int(input("Type here your note between 1 and 100: "))

if note <= 49 :

    print(f"{name} your note was Insufficient")

elif note >= 50 and note <= 69:

    print(f"{name} your note was Satisfactory")

elif note >= 70 and note <= 89:

    print(f"{name} your note was Good")

else:

    print(f"{name} your note was Excelent")


print("-----------------------------------")
print(f"Student : {name} | Note: {note}")
```

```python
value = int(input("Hello client how what is the value of your purchase: "))

if value > 1000:

    value2 = value - (value * 0.2)
    print("You have 20% off ")
    print(f"Pay {value2}")

elif value >= 500 and value <= 1000:

    value2 = value - (value * 0.1)
    print("You have 10% off ")
    print(f"Pay {value2}")

else:

    print("You don't have any discount")
    print(f"Pay {value}")
```
